```python
# Milestone 2: Recommendation System Model Building (Memory Optimized)
# Objective: Develop and train the core recommendation model
# Optimized for large datasets using sparse matrices

# ============================================================================
# 1. IMPORT LIBRARIES
# ============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import warnings
import gc
warnings.filterwarnings('ignore')

# Sklearn imports
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error

# Scipy sparse matrices for memory efficiency
from scipy.sparse import csr_matrix, lil_matrix
from scipy.spatial.distance import cosine

# Set display options
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

print("Libraries imported successfully!")
print("Using sparse matrices for memory efficiency")

# ============================================================================
# 2. LOAD DATASETS (PARQUET FORMAT) - MEMORY EFFICIENT
# ============================================================================

print("\n" + "=" * 80)
print("LOADING DATASETS")
print("=" * 80)

# Load datasets with only necessary columns to save memory
print("Loading products...")
products_df = pd.read_parquet('products0.parquet')

print("Loading users...")
users_df = pd.read_parquet('users0.parquet')

print("Loading interactions...")
interactions_df = pd.read_parquet('interactions_part_0.parquet')

print(f"\n✓ Datasets loaded successfully")
print(f"Products: {products_df.shape}")
print(f"Users: {users_df.shape}")
print(f"Interactions: {interactions_df.shape}")

# Display dataset previews
print("\n--- Products Dataset Sample ---")
print(products_df.head(3))

print("\n--- Users Dataset Sample ---")
print(users_df.head(3))

print("\n--- Interactions Dataset Sample ---")
print(interactions_df.head(3))

# ============================================================================
# 3. DATA PREPROCESSING (MEMORY OPTIMIZED)
# ============================================================================

print("\n" + "=" * 80)
print("DATA PREPROCESSING")
print("=" * 80)

# Clean column names
products_df.columns = products_df.columns.str.strip()
users_df.columns = users_df.columns.str.strip()
interactions_df.columns = interactions_df.columns.str.strip()

# Check for missing values
print("\n--- Missing Values ---")
print(f"Products: {products_df.isnull().sum().sum()}")
print(f"Users: {users_df.isnull().sum().sum()}")
print(f"Interactions: {interactions_df.isnull().sum().sum()}")

# Create interaction scores
def calculate_interaction_score(row):
    """Convert interaction types to numerical scores"""
    if pd.notna(row.get('rating')):
        return float(row['rating'])
    
    interaction_weights = {
        'purchase': 5.0,
        'add_to_cart': 3.0,
        'view': 1.0,
        'click': 1.5,
        'wishlist': 2.5
    }
    
    interaction_type = str(row.get('interaction_type', '')).lower()
    return interaction_weights.get(interaction_type, 1.0)

print("\nCalculating interaction scores...")
interactions_df['score'] = interactions_df.apply(calculate_interaction_score, axis=1)

print("\n--- Interaction Score Distribution ---")
print(interactions_df['score'].describe())

# Remove duplicates - keep highest score
print("\nRemoving duplicates...")
interactions_df = interactions_df.sort_values('score', ascending=False).drop_duplicates(
    subset=['user_id', 'product_id'], keep='first'
).reset_index(drop=True)

print(f"Interactions after deduplication: {len(interactions_df):,}")

# Free memory
gc.collect()

# ============================================================================
# 4. EXPLORATORY DATA ANALYSIS
# ============================================================================

print("\n" + "=" * 80)
print("EXPLORATORY DATA ANALYSIS")
print("=" * 80)

unique_users = interactions_df['user_id'].nunique()
unique_products = interactions_df['product_id'].nunique()
total_interactions = len(interactions_df)

print(f"\nUnique Users: {unique_users:,}")
print(f"Unique Products: {unique_products:,}")
print(f"Total Interactions: {total_interactions:,}")
print(f"Sparsity: {1 - (total_interactions / (unique_users * unique_products)):.6f}")

# Interactions per user
interactions_per_user = interactions_df.groupby('user_id').size()
print(f"\n--- Interactions per User ---")
print(f"Mean: {interactions_per_user.mean():.2f}")
print(f"Median: {interactions_per_user.median():.2f}")
print(f"Min: {interactions_per_user.min()}")
print(f"Max: {interactions_per_user.max()}")

# Interactions per product
interactions_per_product = interactions_df.groupby('product_id').size()
print(f"\n--- Interactions per Product ---")
print(f"Mean: {interactions_per_product.mean():.2f}")
print(f"Median: {interactions_per_product.median():.2f}")
print(f"Min: {interactions_per_product.min()}")
print(f"Max: {interactions_per_product.max()}")

# Visualizations
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

axes[0, 0].hist(interactions_per_user.clip(upper=100), bins=50, edgecolor='black', alpha=0.7, color='steelblue')
axes[0, 0].set_title('Distribution of Interactions per User (capped at 100)', fontsize=12, fontweight='bold')
axes[0, 0].set_xlabel('Number of Interactions')
axes[0, 0].set_ylabel('Frequency')
axes[0, 0].grid(axis='y', alpha=0.3)

axes[0, 1].hist(interactions_per_product.clip(upper=100), bins=50, edgecolor='black', alpha=0.7, color='coral')
axes[0, 1].set_title('Distribution of Interactions per Product (capped at 100)', fontsize=12, fontweight='bold')
axes[0, 1].set_xlabel('Number of Interactions')
axes[0, 1].set_ylabel('Frequency')
axes[0, 1].grid(axis='y', alpha=0.3)

axes[1, 0].hist(interactions_df['score'], bins=20, edgecolor='black', alpha=0.7, color='green')
axes[1, 0].set_title('Distribution of Interaction Scores', fontsize=12, fontweight='bold')
axes[1, 0].set_xlabel('Score')
axes[1, 0].set_ylabel('Frequency')
axes[1, 0].grid(axis='y', alpha=0.3)

top_products = interactions_per_product.nlargest(10)
axes[1, 1].barh(range(len(top_products)), top_products.values, color='purple', alpha=0.7)
axes[1, 1].set_yticks(range(len(top_products)))
axes[1, 1].set_yticklabels([f'Prod {pid}' for pid in top_products.index])
axes[1, 1].set_title('Top 10 Products by Interactions', fontsize=12, fontweight='bold')
axes[1, 1].set_xlabel('Number of Interactions')
axes[1, 1].grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.show()

# ============================================================================
# 5. PREPARE DATA FOR MODELING (SPARSE MATRICES)
# ============================================================================

print("\n" + "=" * 80)
print("PREPARING DATA FOR MODELING")
print("=" * 80)

# Filter to reduce matrix size - keep active users and popular products
min_user_interactions = 5
min_product_interactions = 5

user_counts = interactions_df['user_id'].value_counts()
product_counts = interactions_df['product_id'].value_counts()

valid_users = user_counts[user_counts >= min_user_interactions].index
valid_products = product_counts[product_counts >= min_product_interactions].index

interactions_filtered = interactions_df[
    (interactions_df['user_id'].isin(valid_users)) & 
    (interactions_df['product_id'].isin(valid_products))
].copy()

print(f"\nFiltered interactions: {len(interactions_filtered):,} (from {len(interactions_df):,})")
print(f"Active users: {interactions_filtered['user_id'].nunique():,}")
print(f"Popular products: {interactions_filtered['product_id'].nunique():,}")

# Create train-test split
train_data, test_data = train_test_split(
    interactions_filtered, 
    test_size=0.2, 
    random_state=42
)

print(f"\nTraining set: {len(train_data):,}")
print(f"Test set: {len(test_data):,}")

# Create mappings for sparse matrix indices
print("\nCreating user and product mappings...")
unique_users_list = sorted(train_data['user_id'].unique())
unique_products_list = sorted(train_data['product_id'].unique())

user_to_idx = {user: idx for idx, user in enumerate(unique_users_list)}
idx_to_user = {idx: user for user, idx in user_to_idx.items()}

product_to_idx = {prod: idx for idx, prod in enumerate(unique_products_list)}
idx_to_product = {idx: prod for prod, idx in product_to_idx.items()}

n_users = len(unique_users_list)
n_products = len(unique_products_list)

print(f"Matrix dimensions: {n_users:,} users × {n_products:,} products")

# Create sparse user-item matrix
print("\nCreating sparse user-item matrix...")
row_indices = train_data['user_id'].map(user_to_idx).values
col_indices = train_data['product_id'].map(product_to_idx).values
scores = train_data['score'].values

train_matrix_sparse = csr_matrix(
    (scores, (row_indices, col_indices)), 
    shape=(n_users, n_products)
)

print(f"Sparse matrix created: {train_matrix_sparse.shape}")
print(f"Non-zero elements: {train_matrix_sparse.nnz:,}")
print(f"Sparsity: {1 - (train_matrix_sparse.nnz / (n_users * n_products)):.6f}")
print(f"Memory usage: ~{train_matrix_sparse.data.nbytes / 1024**2:.2f} MB")

gc.collect()

# ============================================================================
# 6. MODEL 1: ITEM-BASED COLLABORATIVE FILTERING (MEMORY EFFICIENT)
# ============================================================================

print("\n" + "=" * 80)
print("MODEL 1: ITEM-BASED COLLABORATIVE FILTERING")
print("=" * 80)

print("\nNote: Using item-based CF as it's more memory efficient for this dataset")
print("(More users than products)")

# Compute item similarity for popular items only
top_n_items = 1000  # Limit similarity computation
print(f"\nComputing similarity for top {top_n_items} most popular items...")

popular_items_idx = np.array(train_matrix_sparse.getnnz(axis=0).argsort()[-top_n_items:][::-1])
print(f"Selected {len(popular_items_idx)} items for similarity computation")

# Store similarities in a dictionary (sparse storage)
item_similarities = {}

def compute_item_similarity(idx1, idx2):
    """Compute cosine similarity between two items"""
    vec1 = train_matrix_sparse[:, idx1].toarray().flatten()
    vec2 = train_matrix_sparse[:, idx2].toarray().flatten()
    
    if np.sum(vec1) == 0 or np.sum(vec2) == 0:
        return 0.0
    
    return 1 - cosine(vec1, vec2)

# Compute similarities for popular items
print("Computing item-item similarities...")
for i, item_idx in enumerate(popular_items_idx):
    if i % 100 == 0:
        print(f"  Processed {i}/{len(popular_items_idx)} items...")
    
    item_similarities[item_idx] = {}
    
    # Only compute with other popular items
    for other_idx in popular_items_idx:
        if item_idx != other_idx:
            sim = compute_item_similarity(item_idx, other_idx)
            if sim > 0.1:  # Only store significant similarities
                item_similarities[item_idx][other_idx] = sim

print(f"✓ Item similarities computed")

def predict_item_based_sparse(user_id, product_id, k=20):
    """Predict rating using item-based CF with sparse matrices"""
    
    if user_id not in user_to_idx or product_id not in product_to_idx:
        return train_data['score'].mean()
    
    user_idx = user_to_idx[user_id]
    product_idx = product_to_idx[product_id]
    
    # Get user's rated items
    user_ratings = train_matrix_sparse[user_idx].toarray().flatten()
    rated_items = np.where(user_ratings > 0)[0]
    
    if len(rated_items) == 0:
        col_mean = train_matrix_sparse[:, product_idx].toarray().flatten()
        return col_mean[col_mean > 0].mean() if np.any(col_mean > 0) else train_data['score'].mean()
    
    # Get similar items
    if product_idx not in item_similarities:
        col_mean = train_matrix_sparse[:, product_idx].toarray().flatten()
        return col_mean[col_mean > 0].mean() if np.any(col_mean > 0) else train_data['score'].mean()
    
    similar_items = item_similarities[product_idx]
    
    ratings = []
    weights = []
    
    for item_idx in rated_items:
        if item_idx in similar_items:
            sim = similar_items[item_idx]
            if sim > 0:
                ratings.append(user_ratings[item_idx])
                weights.append(sim)
    
    if not ratings:
        col_mean = train_matrix_sparse[:, product_idx].toarray().flatten()
        return col_mean[col_mean > 0].mean() if np.any(col_mean > 0) else train_data['score'].mean()
    
    # Take top k most similar
    if len(ratings) > k:
        top_k_idx = np.argsort(weights)[-k:]
        ratings = [ratings[i] for i in top_k_idx]
        weights = [weights[i] for i in top_k_idx]
    
    return np.average(ratings, weights=weights)

# Evaluate item-based CF on a sample of test data
print("\nEvaluating Item-Based CF on test set...")
sample_size = min(5000, len(test_data))
test_sample = test_data.sample(n=sample_size, random_state=42)

item_based_predictions = []
item_based_actuals = []

for idx, row in test_sample.iterrows():
    pred = predict_item_based_sparse(row['user_id'], row['product_id'])
    item_based_predictions.append(pred)
    item_based_actuals.append(row['score'])

item_based_rmse = np.sqrt(mean_squared_error(item_based_actuals, item_based_predictions))
item_based_mae = mean_absolute_error(item_based_actuals, item_based_predictions)

print(f"\nItem-Based CF Performance (on {sample_size} test samples):")
print(f"RMSE: {item_based_rmse:.4f}")
print(f"MAE: {item_based_mae:.4f}")

gc.collect()

# ============================================================================
# 7. MODEL 2: MATRIX FACTORIZATION (ALS - MEMORY EFFICIENT)
# ============================================================================

print("\n" + "=" * 80)
print("MODEL 2: MATRIX FACTORIZATION (ALS)")
print("=" * 80)

class AlternatingLeastSquares:
    """Simple ALS implementation for memory efficiency"""
    
    def __init__(self, n_factors=20, n_iterations=10, reg_param=0.1):
        self.n_factors = n_factors
        self.n_iterations = n_iterations
        self.reg_param = reg_param
        
    def fit(self, sparse_matrix):
        """Fit the model using alternating least squares"""
        n_users, n_items = sparse_matrix.shape
        
        # Initialize factors
        print(f"\nInitializing {self.n_factors} latent factors...")
        self.user_factors = np.random.normal(0, 0.1, (n_users, self.n_factors))
        self.item_factors = np.random.normal(0, 0.1, (n_items, self.n_factors))
        
        print(f"Training ALS for {self.n_iterations} iterations...")
        for iteration in range(self.n_iterations):
            # Update user factors
            for u in range(n_users):
                items = sparse_matrix[u].nonzero()[1]
                if len(items) > 0:
                    A = self.item_factors[items].T @ self.item_factors[items] + \
                        self.reg_param * np.eye(self.n_factors)
                    b = self.item_factors[items].T @ sparse_matrix[u, items].toarray().flatten()
                    self.user_factors[u] = np.linalg.solve(A, b)
            
            # Update item factors
            for i in range(n_items):
                users = sparse_matrix[:, i].nonzero()[0]
                if len(users) > 0:
                    A = self.user_factors[users].T @ self.user_factors[users] + \
                        self.reg_param * np.eye(self.n_factors)
                    b = self.user_factors[users].T @ sparse_matrix[users, i].toarray().flatten()
                    self.item_factors[i] = np.linalg.solve(A, b)
            
            print(f"  Iteration {iteration + 1}/{self.n_iterations} completed")
        
        print("✓ ALS training completed")
    
    def predict(self, user_idx, item_idx):
        """Predict rating for user-item pair"""
        return self.user_factors[user_idx] @ self.item_factors[item_idx]

# Train ALS model
als_model = AlternatingLeastSquares(n_factors=15, n_iterations=5, reg_param=0.1)
als_model.fit(train_matrix_sparse)

def predict_als(user_id, product_id):
    """Predict rating using ALS"""
    if user_id not in user_to_idx or product_id not in product_to_idx:
        return train_data['score'].mean()
    
    user_idx = user_to_idx[user_id]
    product_idx = product_to_idx[product_id]
    
    prediction = als_model.predict(user_idx, product_idx)
    
    # Clip to valid range
    min_score = train_data['score'].min()
    max_score = train_data['score'].max()
    return np.clip(prediction, min_score, max_score)

# Evaluate ALS
print("\nEvaluating ALS on test set...")
als_predictions = []
als_actuals = []

for idx, row in test_sample.iterrows():
    pred = predict_als(row['user_id'], row['product_id'])
    als_predictions.append(pred)
    als_actuals.append(row['score'])

als_rmse = np.sqrt(mean_squared_error(als_actuals, als_predictions))
als_mae = mean_absolute_error(als_actuals, als_predictions)

print(f"\nALS Performance (on {sample_size} test samples):")
print(f"RMSE: {als_rmse:.4f}")
print(f"MAE: {als_mae:.4f}")

gc.collect()

# ============================================================================
# 8. MODEL 3: POPULARITY-BASED BASELINE
# ============================================================================

print("\n" + "=" * 80)
print("MODEL 3: POPULARITY-BASED BASELINE")
print("=" * 80)

# Calculate product popularity scores
product_popularity = train_data.groupby('product_id').agg({
    'score': ['mean', 'count']
}).reset_index()
product_popularity.columns = ['product_id', 'avg_score', 'interaction_count']

scaler = MinMaxScaler()
product_popularity['normalized_count'] = scaler.fit_transform(
    product_popularity[['interaction_count']]
)
product_popularity['popularity_score'] = (
    product_popularity['avg_score'] * 0.7 + 
    product_popularity['normalized_count'] * 5 * 0.3
)

print("\nTop 10 Popular Products:")
print(product_popularity.nlargest(10, 'popularity_score')[
    ['product_id', 'avg_score', 'interaction_count', 'popularity_score']
].to_string(index=False))

# Evaluate popularity-based
popularity_predictions = []
popularity_actuals = []

global_mean = train_data['score'].mean()

print("\nEvaluating Popularity-Based model...")
for idx, row in test_sample.iterrows():
    pop_score = product_popularity[
        product_popularity['product_id'] == row['product_id']
    ]['popularity_score'].values
    
    pred = pop_score[0] if len(pop_score) > 0 else global_mean
    popularity_predictions.append(pred)
    popularity_actuals.append(row['score'])

popularity_rmse = np.sqrt(mean_squared_error(popularity_actuals, popularity_predictions))
popularity_mae = mean_absolute_error(popularity_actuals, popularity_predictions)

print(f"\nPopularity-Based Performance:")
print(f"RMSE: {popularity_rmse:.4f}")
print(f"MAE: {popularity_mae:.4f}")

# ============================================================================
# 9. MODEL 4: HYBRID MODEL
# ============================================================================

print("\n" + "=" * 80)
print("MODEL 4: HYBRID MODEL")
print("=" * 80)

def predict_hybrid(user_id, product_id, weights={'item': 0.5, 'als': 0.4, 'pop': 0.1}):
    """Hybrid prediction combining all models"""
    predictions = []
    
    # Item-based CF
    pred_item = predict_item_based_sparse(user_id, product_id)
    predictions.append(pred_item * weights['item'])
    
    # ALS
    pred_als = predict_als(user_id, product_id)
    predictions.append(pred_als * weights['als'])
    
    # Popularity
    pop_score = product_popularity[
        product_popularity['product_id'] == product_id
    ]['popularity_score'].values
    pred_pop = pop_score[0] if len(pop_score) > 0 else global_mean
    predictions.append(pred_pop * weights['pop'])
    
    return sum(predictions)

# Evaluate hybrid model
print("\nEvaluating Hybrid model...")
hybrid_predictions = []
hybrid_actuals = []

for idx, row in test_sample.iterrows():
    pred = predict_hybrid(row['user_id'], row['product_id'])
    hybrid_predictions.append(pred)
    hybrid_actuals.append(row['score'])

hybrid_rmse = np.sqrt(mean_squared_error(hybrid_actuals, hybrid_predictions))
hybrid_mae = mean_absolute_error(hybrid_actuals, hybrid_predictions)

print(f"\nHybrid Model Performance:")
print(f"RMSE: {hybrid_rmse:.4f}")
print(f"MAE: {hybrid_mae:.4f}")

# ============================================================================
# 10. PERFORMANCE COMPARISON
# ============================================================================

print("\n" + "=" * 80)
print("MODEL PERFORMANCE COMPARISON")
print("=" * 80)

results = {
    'Model': [
        'Item-Based CF',
        'ALS (Matrix Factorization)',
        'Popularity-Based',
        'Hybrid Model'
    ],
    'RMSE': [
        item_based_rmse,
        als_rmse,
        popularity_rmse,
        hybrid_rmse
    ],
    'MAE': [
        item_based_mae,
        als_mae,
        popularity_mae,
        hybrid_mae
    ]
}

results_df = pd.DataFrame(results)
results_df = results_df.sort_values('RMSE')
results_df['Rank'] = range(1, len(results_df) + 1)

print(f"\nEvaluated on {sample_size:,} test samples:")
print("\n" + results_df.to_string(index=False))

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

x = range(len(results_df))
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#9467bd']

axes[0].bar(x, results_df['RMSE'], color=colors, alpha=0.7, edgecolor='black')
axes[0].set_xticks(x)
axes[0].set_xticklabels(results_df['Model'], rotation=45, ha='right')
axes[0].set_ylabel('RMSE', fontweight='bold')
axes[0].set_title('Model Comparison - RMSE (Lower is Better)', fontsize=12, fontweight='bold')
axes[0].grid(axis='y', alpha=0.3)
for i, v in enumerate(results_df['RMSE']):
    axes[0].text(i, v + 0.02, f'{v:.3f}', ha='center', va='bottom', fontweight='bold', fontsize=9)

axes[1].bar(x, results_df['MAE'], color=colors, alpha=0.7, edgecolor='black')
axes[1].set_xticks(x)
axes[1].set_xticklabels(results_df['Model'], rotation=45, ha='right')
axes[1].set_ylabel('MAE', fontweight='bold')
axes[1].set_title('Model Comparison - MAE (Lower is Better)', fontsize=12, fontweight='bold')
axes[1].grid(axis='y', alpha=0.3)
for i, v in enumerate(results_df['MAE']):
    axes[1].text(i, v + 0.02, f'{v:.3f}', ha='center', va='bottom', fontweight='bold', fontsize=9)

plt.tight_layout()
plt.show()

# ============================================================================
# 11. GENERATE RECOMMENDATIONS
# ============================================================================

print("\n" + "=" * 80)
print("GENERATING SAMPLE RECOMMENDATIONS")
print("=" * 80)

def get_recommendations(user_id, n=10, method='hybrid'):
    """Generate top N recommendations for a user"""
    
    if user_id not in user_to_idx:
        # For new users, recommend popular items
        return product_popularity.nlargest(n, 'popularity_score')['product_id'].tolist()
    
    user_idx = user_to_idx[user_id]
    
    # Get products user hasn't interacted with
    user_rated = train_matrix_sparse[user_idx].nonzero()[1]
    user_products = set([idx_to_product[idx] for idx in user_rated])
    all_products = set(unique_products_list)
    candidates = list(all_products - user_products)
    
    # Limit candidates for efficiency
    if len(candidates) > 500:
        # Sample candidates based on popularity
        pop_candidates = product_popularity[
            product_popularity['product_id'].isin(candidates)
        ].nlargest(500, 'popularity_score')['product_id'].tolist()
        candidates = pop_candidates
    
    # Predict scores
    predictions = []
    for product in candidates:
        if method == 'item_based':
            score = predict_item_based_sparse(user_id, product)
        elif method == 'als':
            score = predict_als(user_id, product)
        elif method == 'hybrid':
            score = predict_hybrid(user_id, product)
        else:
            score = predict_hybrid(user_id, product)
        
        predictions.append((product, score))
    
    # Sort and return top N
    predictions.sort(key=lambda x: x[1], reverse=True)
    return [p[0] for p in predictions[:n]]

# Generate recommendations for sample users
sample_users = unique_users_list[:5]

print(f"\nGenerating recommendations for {len(sample_users)} sample users...\n")

for user in sample_users:
    print(f"{'='*60}")
    print(f"User ID: {user}")
    print(f"{'='*60}")
    
    # Get user's past interactions
    user_history = train_data[train_data['user_id'] == user]
    print(f"Past Interactions: {len(user_history)}")
    
    # Generate recommendations
    recommendations = get_recommendations(user, n=5, method='hybrid')
    
    print("\nTop 5 Recommended Products:")
    for i, prod_id in enumerate(recommendations, 1):
        prod_info = products_df[products_df['product_id'] == prod_id]
        if not prod_info.empty:
            prod_name = prod_info['product_name'].values[0]
            prod_cat = prod_info['category'].values[0] if 'category' in prod_info.columns else 'N/A'
            print(f"  {i}. Product ID: {prod_id}")
            print(f"     Name: {prod_name}")
            print(f"     Category: {prod_cat}")
        else:
            print(f"  {i}. Product ID: {prod_id}")
        print()

# ============================================================================
# 12. SAVE RESULTS
# ============================================================================

print("\n" + "=" * 80)
print("SAVING RESULTS")
print("=" * 80)

# Save performance results
results_df.to_csv('model_performance_results.csv', index=False)
print("✓ Performance results saved to: model_performance_results.csv")

# Save product popularity
product_popularity.to_csv('product_popularity_scores.csv', index=False)
print("✓ Product popularity scores saved")

print("\n" + "=" * 80)
print("MILESTONE 2 COMPLETED SUCCESSFULLY!")
print("=" * 80)
print("\nKey Achievements:")
print("✓ Memory-efficient implementation using sparse matrices")
print("✓ Multiple recommendation algorithms implemented")
print("✓ Models trained and evaluated")
print("✓ Performance benchmarks established")
print("✓ Recommendation generation demonstrated")
print(f"\nBest performing model: {results_df.iloc[0]['Model']}")
```
